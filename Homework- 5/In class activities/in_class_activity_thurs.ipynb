## How to set up hidden layers?
- Funnel: 7 --> 5 --> 3 (good if you have tabular data)
- Flat: 7 --> 7 --> 7 (same number of neurons in every layer)
- Expanding: 3 --> 5 --> 7 
- Non-monotonic 7 --> 5 --> 7 (lot of layers and neurons)

## Network capacity
- 4 --> 2 --> 3 (small, singular numbers)
- 16 --> 8 --> 3 (medium, double digits)
- 100, 90, 80 (large)

## How to determine best capacity?
- 1-2 hidden layers 
- 8-64 neurons per layer
- funnel shape, e.g. 32 --> 16 or 16 --> 8 
- you can continue to drop the neurons, but the loss is representative in the validation test set
