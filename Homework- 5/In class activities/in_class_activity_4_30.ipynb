## How to make the model make corrections and scale them down?
- the learning rate 
- if the learning rate is 0.1, this doesn't correct is quite as far 
- look at the values and think about how much needs to be updated
- next 256 values in mini batches and passed through the learning rate 
- first the mini batches of data are used and weights are corrected
- shuffle data in between epochs
- weights and biases can be adjusted and fixed multiple times during an epoch

## Backpropogation 
- difference in the loss relative to the weight 
- in the second layer, math performed in between layers --> there is loss contributed in the output layer and the proceeding layer
- moving back to layer 1, how did the previous layers impact the current one 

## Vanishing gradient problem
- For each layer there is an activation function, ending up with a number less than 1 
- sigmoid function will keep showing up in subsequent steps: largest gradient problem
- gradient values get smaller and smaller the further down the layers you go
