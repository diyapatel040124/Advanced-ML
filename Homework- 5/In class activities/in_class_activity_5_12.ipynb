## Keras
- Create model --> model.compile --> model.fit(), model.evaluate(), model.predict()
- Dataset (training data, testing data) --> model.fit(), model.evaluate()
1. Define input layers (2 features --> 2 input nodes)
2. first hidden layer (2 neurons, ReLU activation)
3. Second hidden layer (2 neurons, ReLU activation)
4. output layer (1 neuron --> final prediction)
5. Build the final model
