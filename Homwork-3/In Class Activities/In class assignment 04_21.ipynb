## Box Cox 
- lambda has an impact of how that feature is transformed, there can be a whole different set of transformations that can be applied here 
- values defined must be greater than 0
- ask lambda to search
- instead of manually trying different values of lambda, box cox can find an optimal value for lambda
- negatives/zeros: Yeo-Johnson, most robust because it will give the normal distribution
- this doesn't always improve the distribution of the model, takes up lots of computing power 

## Box Cox/Yeo Johnson to simple transformations 
- 1: no change --> leave it 
- <-1: extreme compression, reconsider/use Box Cox 

## When to transform?
- If using linear, SVM, Neaural Networks, and KNN models would be the ones to transforms 
- they use distance calculations as a part of how they handle the data 

## Encoding categorical features 
- consider combining rare categories 
- one hot encoding --> this is the default, generates sparse data and has many features 
- ordinal encoding --> only use if there is true order to the data, model assumes the distance between levels is meaningful
- target encoding --> steps: 
    - split (use folds, out of fold averages --> averages of train. Never encode with observation's own target)
- tree-based model will be better on making predictions 
- combining categories can be a good option
- XGBoost and CatBoost are good for these 

## Imputing Missing Values 
- "Missingness" 
- Simple: 
    - Numerical mean or median, categorical: "unknown"
- train a model to predict it 
    - remove target before imputing
    - train test split --> train only with those with a value --> impute train/missing test
